### <a name='a0'></a> import biblioteki

```python
import pandas as pd
import numpy as np
import yfinance as yf
```

### <a name='a1'></a> Załadowanie danych

```python
# funkcja ktora pobiera dane z serwisu yahoo finance
def fetch_financial_data(company='FB'):
    """Pobiera dane dotyczące notowań spółki."""
    return yf.download(company)

df = fetch_financial_data()

df.head()
```

```python
# spłaszczenie kolumny w yfinance
df.columns = df.columns.get_level_values(0)
```

### <a name='a2'></a> Zapis/Odczyt z pliku danych

```python
# wygenerowanie pliku roboczego dla naszych danych
df.to_csv('fb.csv')
# wyeksportowany plik znajduje sie w zakladkach files
```

```python
# dane z danego miesiaca
df_nov = df[(df.index.month == 7) & (df.index.year == 2026)]
df_nov
```

```python
# zapisanie danych do pliku
df_nov.to_csv('fb_nov.csv')
```

```python
# wczytanie danych z pliku
new_df = pd.read_csv('fb_nov.csv')
new_df
```

```python
# zapis danych do formatu pliku excel
df_nov.to_excel('fb_nov.xlsx')
```

```python
# wczytanie danych z pliku o fromacie excel
new_df = pd.read_excel('fb_nov.xlsx')
new_df
# ale uwaga data została wczytana jako kolumna a nie sam w sobie index jak poprzednio
```

```python
# dlatego musimy sobie ustawic kolumne daty jako index poprzez parametr index
new_df = pd.read_excel('fb_nov.xlsx' , index_col=0)
# data staje sie indeksem naszej tabeli
new_df
```
